# Question 2 Take two numbers from the user and perform
## ● Addition
## ● Subtraction
## ● Multiplication
## ● Division
## ● Floor Division
## ● Modulus
## ● Exponent

```python
# Take two numbers from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Perform operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2

# Handle division safely
if num2 != 0:
    division = num1 / num2
    floor_division = num1 // num2
    modulus = num1 % num2
else:
    division = "Error: Division by zero"
    floor_division = "Error: Division by zero"
    modulus = "Error: Division by zero"

# Exponentiation
exponent = num1 ** num2

# Print results using formatted strings
print(f"Addition: {num1} + {num2} = {addition}")
print(f"Subtraction: {num1} - {num2} = {subtraction}")
print(f"Multiplication: {num1} * {num2} = {multiplication}")
print(f"Division: {num1} / {num2} = {division}")
print(f"Floor Division: {num1} // {num2} = {floor_division}")
print(f"Modulus: {num1} % {num2} = {modulus}")
print(f"Exponent: {num1} ** {num2} = {exponent}")
```

# Question 3 Write a program to check whether a student has passed.
## Conditions
## ● Marks ≥ 90 → Grade A
## ● Marks ≥ 75 → Grade B
## ● Marks ≥ 60 → Grade C
## ● Marks ≥ 40 → Grade D
## ● Otherwise Fail

```python
# Take marks from the user
marks = int(input("Enter the student's marks: "))

# Check grade conditions
if marks >= 90:
    grade = "A"
    message = "Excellent! Congratulations, you passed with Grade A."
elif marks >= 75:
    grade = "B"
    message = "Great job! You passed with Grade B."
elif marks >= 60:
    grade = "C"
    message = "Good work! You passed with Grade C."
elif marks >= 40:
    grade = "D"
    message = "You passed with Grade D. Keep improving!"
else:
    grade = "Fail"
    message = "Sorry, you failed. Better luck next time."

# Display result
print(f"Marks: {marks} → Grade: {grade}")
print(message)
```

# Question 6 Create a dictionary of five students.
## Store
### ● Name
### ● Age
### ● Marks
## Perform
### ● Add new student
### ● Delete one student
### ● Update marks
### ● Print all students

```python
# Create a dictionary of five students
students = {
    1: {"Name": "Amit", "Age": 20, "Marks": 85},
    2: {"Name": "Riya", "Age": 21, "Marks": 92},
    3: {"Name": "Karan", "Age": 19, "Marks": 76},
    4: {"Name": "Sneha", "Age": 22, "Marks": 64},
    5: {"Name": "Vikas", "Age": 20, "Marks": 55}
}

# Function to print students in table format
def print_students(data):
    print("\n{:<5} {:<10} {:<5} {:<6}".format("ID", "Name", "Age", "Marks"))
    print("-" * 30)
    for sid, info in data.items():
        print("{:<5} {:<10} {:<5} {:<6}".format(sid, info['Name'], info['Age'], info['Marks']))

# Original list
print("Original Students Dictionary:")
print_students(students)

# Add new student
students[6] = {"Name": "Meera", "Age": 21, "Marks": 88}
print("\nAfter Adding New Student:")
print_students(students)

# Delete one student
del students[3]   # Delete student with ID 3
print("\nAfter Deleting Student with ID 3:")
print_students(students)

# Update marks
students[2]["Marks"] = 95   # Update Riya's marks
print("\nAfter Updating Marks of Student with ID 2:")
print_students(students)

# Final list
print("\nFinal Students Dictionary:")
print_students(students)
```

# Question 7 Write a user-defined function
## calculate_area(length,width)
## Return the area.
## Also create another function
## calculate_perimeter(length,width)

```python
# Function to calculate area
def calculate_area(length, width):
    return length * width

# Function to calculate perimeter
def calculate_perimeter(length, width):
    return 2 * (length + width)

# Example usage
l = float(input("Enter the length: "))
w = float(input("Enter the width: "))

area = calculate_area(l, w)
perimeter = calculate_perimeter(l, w)

print(f"Area of rectangle = {area}")
print(f"Perimeter of rectangle = {perimeter}")
```

# Question 14 Given
## [10,20,30,40,50,60]
## Perform
### ● Reshape
### ● Transpose
### ● Sort
### ● Reverse
### ● Split into two arrays

```python
import numpy as np

# Original array
arr = np.array([10, 20, 30, 40, 50, 60])
print("Original Array:", arr)

# Reshape (2D: 2x3 matrix)
reshaped = arr.reshape(2, 3)
print("\nReshaped Array (2x3):\n", reshaped)

# Transpose
transposed = reshaped.T
print("\nTransposed Array:\n", transposed)

# Sort
sorted_arr = np.sort(arr)
print("\nSorted Array:", sorted_arr)

# Reverse
reversed_arr = arr[::-1]
print("\nReversed Array:", reversed_arr)

# Split into two arrays
split_arr = np.array_split(arr, 2)
print("\nSplit Arrays:")
print("First Half:", split_arr[0])
print("Second Half:", split_arr[1])
```

# Section F: Data Visualization
## Create the following charts.
## 1. Line Plot
## 2. Bar Plot
## 3. Scatter Plot
## 4. Histogram
## 5. Pie Chart
## 6. Boxplot
## 7. Heatmap
# For each graph
## ● Write one observation.
## ● Mention what insights can be derived.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create sample dataset (all columns length = 50)
np.random.seed(42)
df = pd.DataFrame({
    "Day": np.arange(1, 51),   # 50 days
    "Sales": np.random.randint(50, 200, 50),
    "Category": np.random.choice(["A","B","C","D"], 50),
    "Value": np.random.randint(10, 100, 50),
    "X": np.random.randn(50),
    "Y": np.random.randn(50)*20 + 100,
    "Age": np.random.randint(18, 60, 50)
})

# Create subplots grid
fig, axes = plt.subplots(4, 2, figsize=(14, 16))
axes = axes.flatten()

# 1. Line Plot
axes[0].plot(df["Day"], df["Sales"], marker="o")
axes[0].set_title("Line Plot - Sales over Days")
axes[0].set_xlabel("Day")
axes[0].set_ylabel("Sales")

# 2. Bar Plot
df["Category"].value_counts().plot(kind="bar", color="skyblue", ax=axes[1])
axes[1].set_title("Bar Plot - Category Counts")

# 3. Scatter Plot
axes[2].scatter(df["X"], df["Y"], alpha=0.7, color="green")
axes[2].set_title("Scatter Plot - X vs Y")
axes[2].set_xlabel("X")
axes[2].set_ylabel("Y")

# 4. Histogram
sns.histplot(df["Age"], bins=10, kde=True, color="salmon", ax=axes[3])
axes[3].set_title("Histogram - Age Distribution")

# 5. Pie Chart
sizes = df["Category"].value_counts()
axes[4].pie(sizes, labels=sizes.index, autopct="%1.1f%%", 
            colors=["#ff9999","#66b3ff","#99ff99","#ffcc99"])
axes[4].set_title("Pie Chart - Category Share")

# 6. Boxplot
sns.boxplot(y=df["Value"], ax=axes[5])
axes[5].set_title("Boxplot - Value Distribution")

# 7. Heatmap
corr = df[["Sales","Value","Age","Y"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", ax=axes[6])
axes[6].set_title("Heatmap - Correlation Matrix")

# Hide the last empty subplot
axes[7].axis("off")

plt.tight_layout()
plt.show()
```
