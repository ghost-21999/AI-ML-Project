# 2. Core Python Concepts:
  ○ Write user-defined functions utilizing flexible parameters (*args, **kwargs) to compute custom revenue metrics.

```python
# ==========================================
# Flexible Revenue Metrics Function
# ==========================================

def compute_revenue(*args, **kwargs):
    """
    Compute custom revenue metrics.
    *args: sales values (can be daily, monthly, or store totals)
    **kwargs: optional parameters like discount, tax, bonus
    """
    # Base revenue = sum of all sales passed in
    revenue = sum(args)

    # Apply optional discount
    if "discount" in kwargs:
        revenue *= (1 - kwargs["discount"])

    # Apply optional tax
    if "tax" in kwargs:
        revenue *= (1 + kwargs["tax"])

    # Add optional bonus
    if "bonus" in kwargs:
        revenue += kwargs["bonus"]

    return revenue


# -----------------------------
# Example Usage
# -----------------------------
# Case 1: Simple sum of sales
print("Revenue 1:", compute_revenue(1000, 2000, 1500))

# Case 2: With discount
print("Revenue 2:", compute_revenue(1000, 2000, discount=0.1))

# Case 3: With discount + tax + bonus
print("Revenue 3:", compute_revenue(5000, 3000, 2000, discount=0.05, tax=0.18, bonus=1000))
```

## 2. Control Flow & Exception Handling:
   ○ Implement a robust data importer module wrapped in try-except-else-finally blocks to catch potential FileNotFoundError, KeyError, and ValueError.

```python
import json

def import_store_config(filename):
    """
    Robust data importer for store configuration JSON.
    Handles FileNotFoundError, KeyError, and ValueError.
    """
    try:
        # Attempt to open and load JSON file
        with open(filename, "r") as file:
            data = json.load(file)

        # Access required keys (may raise KeyError)
        metadata = data["metadata"]
        categories = data["categories"]

        # Example validation: ensure category IDs are integers
        for key in categories.keys():
            if not key.isdigit():
                raise ValueError(f"Invalid category ID: {key}")

    except FileNotFoundError:
        print("❌ Error: The file was not found.")
        data = None

    except KeyError as e:
        print(f"❌ Error: Missing expected key in JSON → {e}")
        data = None

    except ValueError as e:
        print(f"❌ Error: Invalid data format → {e}")
        data = None

    else:
        # Executes only if no exception occurs
        print("✅ Data imported successfully.")
        print("Stores available:", list(metadata.keys()))
        print("Categories available:", list(categories.values()))

    finally:
        # Always executes, whether error or success
        print("📌 Import process completed.")

    return data


# -----------------------------
# Example Usage
# -----------------------------
config = import_store_config("store_config.json")
```

### ○ Outlier Detection: Use the Interquartile Range (IQR) method to detect sales outliers and treat or trim them.

```python
import pandas as pd
import numpy as np

# Sample sales data
np.random.seed(42)
df = pd.DataFrame({
    "Store_ID": range(1, 21),
    "Sales": np.random.randint(200, 2000, size=20)  # deliberately wide range
})

print("Original DataFrame:\n", df)

# Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"\nQ1 = {Q1}, Q3 = {Q3}, IQR = {IQR}")
print(f"Lower Bound = {lower_bound}, Upper Bound = {upper_bound}")

# Detect outliers
outliers = df[(df["Sales"] < lower_bound) | (df["Sales"] > upper_bound)]
print("\n🔍 Outliers detected:\n", outliers)

df_trimmed = df[(df["Sales"] >= lower_bound) & (df["Sales"] <= upper_bound)]
print("\n✅ DataFrame after trimming outliers:\n", df_trimmed)

df_capped = df.copy()
df_capped["Sales"] = np.where(df_capped["Sales"] > upper_bound, upper_bound,
                              np.where(df_capped["Sales"] < lower_bound, lower_bound, df_capped["Sales"]))
print("\n✅ DataFrame after treating outliers (capping):\n", df_capped)
```

### ○ Date-Time Functionalities: Convert text date columns into Pandas DatetimeIndex, extract Month, Day, and DayOfWeek, and perform time-based resamplings.

### ○ Aggregation: Perform .groupby() operations to summarize average sales and units sold per category, store location, and month.

```python
import pandas as pd

# Sales transactions
sales_df = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Store_ID": [1, 2, 1, 3, 2, 3],
    "Category": ["Electronics", "Clothing", "Clothing", "Electronics", "Groceries", "Groceries"],
    "Units_Sold": [5, 10, 8, 6, 15, 20],
    "Sales": [2500, 400, 320, 5000, 450, 800],
    "Order_Date": ["2026-01-15", "2026-02-20", "2026-03-05", "2026-03-25", "2026-04-10", "2026-04-15"]
})

# Store metadata
store_df = pd.DataFrame({
    "Store_ID": [1, 2, 3],
    "Location": ["Delhi", "Mumbai", "Chennai"]
})

# Merge on Store_ID
df = pd.merge(sales_df, store_df, on="Store_ID", how="left")

# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Month"] = df["Order_Date"].dt.month

print("\n✅ Unified DataFrame:\n", df)

# Average sales and units sold per Category
category_summary = df.groupby("Category")[["Sales", "Units_Sold"]].mean()
print("\n📊 Average per Category:\n", category_summary)

# Average sales and units sold per Store Location
location_summary = df.groupby("Location")[["Sales", "Units_Sold"]].mean()
print("\n📊 Average per Location:\n", location_summary)

# Average sales and units sold per Month
month_summary = df.groupby("Month")[["Sales", "Units_Sold"]].mean()
print("\n📊 Average per Month:\n", month_summary)
```

### 3. Scatter Plot: Units Sold vs. Discount Percentage (highlighting outliers).

```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Sample Data
df = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Store_ID": [1, 2, 1, 3, 2, 3],
    "Category": ["Electronics", "Clothing", "Clothing", "Electronics", "Groceries", "Groceries"],
    "Units_Sold": [5, 10, 8, 6, 15, 20],
    "Sales": [2500, 400, 320, 5000, 450, 800],
    "Discount_Pct": [0.05, 0.10, 0.20, 0.02, 0.30, 0.50],
    "Order_Date": ["2026-01-15", "2026-02-20", "2026-03-05", "2026-03-25", "2026-04-10", "2026-04-15"]
})

# Step 2: Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Step 3: Scatter Plot (Units Sold vs Discount)
fig, ax = plt.subplots(figsize=(7,5))

ax.scatter(df["Discount_Pct"], df["Units_Sold"], color="blue", alpha=0.7)
ax.set_title("Units Sold vs Discount Percentage")
ax.set_xlabel("Discount Percentage")
ax.set_ylabel("Units Sold")

# Step 4: Highlight Outliers using IQR method
Q1 = df["Units_Sold"].quantile(0.25)
Q3 = df["Units_Sold"].quantile(0.75)
IQR = Q3 - Q1

# Define outlier condition
outliers = df[(df["Units_Sold"] < (Q1 - 1.5*IQR)) | (df["Units_Sold"] > (Q3 + 1.5*IQR))]

# Plot outliers in red
ax.scatter(outliers["Discount_Pct"], outliers["Units_Sold"], color="red", marker="x", s=100, label="Outliers")

ax.legend()
plt.show()
```

### 5. Pie Chart: Market share breakdown across store locations.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Sample Data
df = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Store_ID": [1, 2, 1, 3, 2, 3],
    "Category": ["Electronics", "Clothing", "Clothing", "Electronics", "Groceries", "Groceries"],
    "Units_Sold": [5, 10, 8, 6, 15, 20],
    "Sales": [2500, 400, 320, 5000, 450, 800],
    "Order_Date": ["2026-01-15", "2026-02-20", "2026-03-05", "2026-03-25", "2026-04-10", "2026-04-15"]
})

# Store metadata
store_meta = pd.DataFrame({
    "Store_ID": [1, 2, 3],
    "Location": ["Delhi", "Mumbai", "Chennai"]
})

# Step 2: Merge sales with store metadata
df = pd.merge(df, store_meta, on="Store_ID", how="left")

# Step 3: Aggregate total sales per location
location_sales = df.groupby("Location")["Sales"].sum()

# Step 4: Pie Chart
plt.figure(figsize=(6,6))
plt.pie(location_sales.values,
        labels=location_sales.index,
        autopct="%1.1f%%",
        startangle=90,
        colors=["skyblue", "lightgreen", "salmon"])

plt.title("Market Share Breakdown by Store Location")
plt.show()
```
